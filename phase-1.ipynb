## Numpy array and basics

```python
import numpy as np
```

## Creating array from list

```python
arr_1d = np.array([1,2,3,4,5,6])
print("1D array", arr_1d)

arr_2d = np.array([[1,2,3],[4,5,6]])
print("2D array", arr_2d)
```

## List vs numpy array

```python
py_list = [1, 2, 3]
print("Python List multiplication", py_list * 2)

np_arr = np.array([1, 2, 3]) #element wise multiplication
print("Python numpy array multiplication", np_arr * 2)


import time

start = time.time()
py_list = [i * 2 for i in range(10000000)]
print("\n List operation time:", time.time() - start)

start = time.time()
np_arr = np.arange(10000000) * 2
print("\n Numpy operation time:", time.time() - start)


```

## Creating array from zeroes or ones (Matrices)

```python
zeroes = np.zeros((3, 4))

print("Zeroes Array", zeroes)

ones = np.ones((3, 4))

print("1 Array", ones)

full = np.full((2,2), 7)
print("Constant value array", full)

random = np.random.random((2, 3))
print("Random Array", random)

sequence = np.arange(0, 20, 2)
print("Sequence Array:","\n", sequence)
```

### Vector, Matrices and Tensor

```python
vector = np.array([1,2,3,])
print("Vector: ", vector)

matrices = np.array([[1,2,3],[4,5,6]])
print("Matrices: ", matrices)

tensor = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print("Tensor: ", tensor)
```

### Array Properties

```python
arr  = np.array([[1, 2, 3], [4, 5, 6], [6, 7, 8]])

print("Shape of Array ", arr.shape)
print("Dimension", arr.ndim)
print("Size (total element in array)", arr.size)
print("Data type ", arr.dtype) 
```

### Array Reshaping

```python
arr = np.arange(12)
print("Original Array ", arr)

reshaped = arr.reshape((3, 4))
print("\n Reshaped Array ", reshaped)


flattened = reshaped.flatten()
print("\n Flatten Array ", flattened)

#ravel (returns view, instead of copy)
raveled = reshaped.ravel()
print("\n Ravel Array ", raveled)

#Transpose
transpose = reshaped.T
print("\n Transpose Array ", transpose)
```
