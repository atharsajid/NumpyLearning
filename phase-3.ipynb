```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("==== Zomato sales analysis ==== ")
print("\n Sales Data Shape: ", sales_data.shape)
print("\n Sample Data for 1st three restaurant ", sales_data[:3])
```

```python
#Total Sales per year

sum_year_wise = np.sum(sales_data[:, 1:], axis=0)
print("Total Sales per year: ", sum_year_wise)

print("Total Sales", np.sum(sum_year_wise, axis=0))
```

```python
#Minimum Sales per restaurant

min_sales = np.min(sales_data[:, 1:], axis=1)
print("Minimum Sales per restaurant", min_sales)
```

```python
#Maximum Sales per restaurant

max_sales = np.max(sales_data[:, 1:], axis=1)
print("Maximum Sales per restaurant", max_sales)
```

```python
#Maximum Sales per year

max_sales_per_year = np.max(sales_data[:, 1:], axis=0)
print("Maximum Sales per year", max_sales_per_year)
```

```python
#Average Sales per restaurant

average_sales = np.mean(sales_data[:, 1:], axis=1)

print("Average Sales per restaurant: ", average_sales)
```

```python
#Cumulative Sales per restaurant

cumsum = np.cumsum(sales_data[:, 1:], axis=1)

print("Cumulative Sales per restaurant: ", cumsum)

plt.figure(figsize=(10, 6))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average Cumulative Sales across all restaurant")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.grid()
plt.show()


```
